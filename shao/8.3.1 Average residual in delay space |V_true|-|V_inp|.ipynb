# 1. Compare different inpainting algorithms: $\frac{|V_{\rm data}(\tau)|}{|V_{\rm inpainted}(\tau)|}$

```python
import os

bl_name = f"bl_{bl[0]}_{bl[1]}"
freq_time_name = (
    f"freq_{freq_chans.start}_{freq_chans.stop}_"
    f"time_{time_ints.start}_{time_ints.stop}"
)

save_dir = os.path.join(
    "inpaint_results",
    bl_name,
    freq_time_name,
)
```

```python
import os
import torch
import matplotlib.pyplot as plt


# ==================================================
# 0. Helper: title style
# ==================================================

def add_flag_title(ax, flag_id, flag_name, subtitle, y=1.32):
    """
    Add a two-line title above the axis.

    Line 1: flag information in dark red.
    Line 2: plot type in black.
    """
    ax.text(
        0.5,
        y,
        f"Flag {flag_id}: {flag_name}",
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        color="darkred",
        fontsize=12,
        fontweight="bold",
    )

    ax.text(
        0.5,
        y - 0.12,
        subtitle,
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        color="black",
        fontsize=12,
    )


# ==================================================
# 1. Algorithms to compare
# ==================================================

Algorithm_names = [
    "freq_1d",
    "time_1d_freq_1d",
    "time_freq_2d",
    "time_freq_2d_freq_1d",
]

flag_types_all = [
    ["real"],
    ["narrowband_full_time"],
    ["wideband_full_time"],
    ["ultraband_short_time"],
    ["fullband_short_time"],
    ["narrowband_short_time"],
    ["narrowband_intermittent"],
]


# ==================================================
# 1.1 Plot styles for different algorithms
# ==================================================

linestyles = ["-", "--", "-.", ":"]
markers = ["o", ".", "^", "."]
linewidths = [2.2, 2.0, 2.0, 2.3]

# marker 间隔，可以根据点数调整
markevery = 25


# ==================================================
# 2. Loop over flag cases
# ==================================================

for flag_id, flag_types in enumerate(flag_types_all, start=1):

    flag_name = "+".join(flag_types)
    save_path = os.path.join(save_dir, f"inpaint_{flag_name}.pt")

    result_this_flag = torch.load(save_path, map_location="cpu")

    flags = result_this_flag["flags"].to(device)

    print("Loaded:", save_path)
    print(f"Flag {flag_id}: {flag_name}")
    print("flags shape:", flags.shape)

    # --------------------------------------------------
    # 2.1 Coherent time average for original and flagged data
    # --------------------------------------------------
    avg_times = ba.utils.split_into_groups(
        torch.arange(Ntimes),
        Nelem=50,
    )

    # Original data
    vd_avg = vd.time_average(
        avg_times,
        rephase=True,
        inplace=False,
    )

    # Original data with flagged pixels removed
    vd_flg = vd * ~flags

    vd_flg_avg = vd_flg.time_average(
        avg_times,
        wgts=1 / (flags * 1e5 + 1e0),
        rephase=True,
        inplace=False,
    )

    # ==================================================
    # 3. Delay-ratio spectrum:
    #    |V_data| / |V_inpainted|
    #    plot range: -1500 < tau < 1500 ns
    #    shaded region: 800 < |tau| < 1500 ns
    # ==================================================

    fig, ax = plt.subplots(1, 1, figsize=(10, 5))

    # --------------------------------------------------
    # 3.1 Compute original data delay spectrum
    # --------------------------------------------------
    data_delay = FT1(vd_avg.data[0, 0].mean(0))[:].abs().mean(0)

    # Plot range: -1500 < tau < 1500 ns
    plot_mask = (FT1.freqs > -1500) & (FT1.freqs < 1500)

    tau_plot = FT1.freqs[plot_mask]
    data_delay_plot = data_delay[plot_mask]

    # --------------------------------------------------
    # 3.2 Loop over inpainting algorithms and plot ratio
    # --------------------------------------------------
    for Algorithm_id, Algorithm_name in enumerate(Algorithm_names, start=1):

        inp_y = result_this_flag["inpaint"][Algorithm_name]["inp_y"].to(device)

        vd_inp = vd.copy(copydata=True)
        vd_inp.data = inp_y[None, None]

        vd_inp_avg = vd_inp.time_average(
            avg_times,
            rephase=True,
            inplace=False,
        )

        inp_delay = FT1(vd_inp_avg.data[0, 0].mean(0))[:].abs().mean(0)
        inp_delay_plot = inp_delay[plot_mask]

        # |V_data| / |V_inpainted|
        ratio = data_delay_plot / (inp_delay_plot + 1e-30)

        style_id = Algorithm_id - 1

        ax.plot(
            tau_plot.detach().cpu(),
            ratio.detach().cpu(),
            linestyle=linestyles[style_id],
            marker=markers[style_id],
            markevery=markevery,
            markersize=4,
            linewidth=linewidths[style_id],
            alpha=0.9,
            label=f"Algorithm {Algorithm_id}: {Algorithm_name}",
        )

    # --------------------------------------------------
    # 3.3 Shaded delay regions: 800 < |tau| < 1500 ns
    # --------------------------------------------------
    ax.axvspan(
        -1500,
        -800,
        alpha=0.2,
        label=r"$800 < |\tau| < 1500$ ns",
    )

    ax.axvspan(
        800,
        1500,
        alpha=0.2,
    )

    # --------------------------------------------------
    # 3.4 Plot formatting
    # --------------------------------------------------
    ax.axhline(
        1.0,
        color="k",
        ls=":",
        lw=1,
        label="ratio = 1",
    )

    ax.grid()
    ax.set_xlim(-1500, 1500)

    ax.set_xlabel("Delay [ns]")
    ax.set_ylabel(r"$|V_{\rm data}(\tau)| / |V_{\rm inpainted}(\tau)|$")

    add_flag_title(
        ax,
        flag_id,
        flag_name,
        subtitle=r"Delay Ratio: $|V_{\rm data}|/|V_{\rm inpainted}|$",
        y=1.45,
    )

    ax.legend(
        loc="lower center",
        bbox_to_anchor=(0.5, 1.08),
        ncol=3,
        frameon=False,
        fontsize=9,
    )

    plt.tight_layout()
    plt.show()
```

# 2. Compare 2D with other algorithms: fractional residual $\frac{\left||V_{\rm data}(\tau)| - |V_{\rm inpainted}(\tau)|\right|}{|V_{\rm data}(\tau)|}$

## Preliminary thoughts
1) For `2.narrowband_full_time` and `3.wideband_full_time`, `4.ultraband_short_time`, `5.fullband_short_time`, the pure 2D algorithm performs well for full-time flags, but it still appears to have a larger residual than the other algorithms. 

2) For `["real"]` and `["fullband_short_time"]`, the `freq_1d` algorithm performs worse than the other algorithms at small \(\tau\).

This may not be due to a problem with the algorithm itself. Instead, it may be related to the fact that some full-band frequency rows are completely missing. In this case, when computing \(V(\nu)\) after time averaging, the averaged inpainted data may not correspond exactly one-to-one with the averaged true data.

A similar effect also seems to appear in `7.3.2 Average residual in frequency space |V_true|-|V_inp|.ipynb`.

```python
import os
import torch
import matplotlib.pyplot as plt


# ==================================================
# 0. Helper: title style
# ==================================================

def add_flag_title(ax, flag_id, flag_name, subtitle, y=1.32):
    """
    Add a two-line title above the axis.

    Line 1: flag information in dark red.
    Line 2: plot type in black.
    """
    ax.text(
        0.5,
        y,
        f"Flag {flag_id}: {flag_name}",
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        color="darkred",
        fontsize=12,
        fontweight="bold",
    )

    ax.text(
        0.5,
        y - 0.12,
        subtitle,
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        color="black",
        fontsize=12,
    )


# ==================================================
# 1. Algorithms to compare
# ==================================================

# Remove pure 2D algorithm: "time_freq_2d"
Algorithm_names = [
    "freq_1d",
    "time_1d_freq_1d",
    "time_freq_2d_freq_1d",
]

flag_types_all = [
    ["real"],
    ["narrowband_full_time"],
    ["wideband_full_time"],
    ["ultraband_short_time"],
    ["fullband_short_time"],
    ["narrowband_short_time"],
    ["narrowband_intermittent"],
]


# ==================================================
# 1.1 Plot styles for different algorithms
# ==================================================

linestyles = ["-", "--", ":"]
markers = ["o", ".", "."]
linewidths = [2.2, 2.0, 2.3]
markevery = 25


# ==================================================
# 2. Loop over flag cases
# ==================================================

for flag_id, flag_types in enumerate(flag_types_all, start=1):

    flag_name = "+".join(flag_types)
    save_path = os.path.join(save_dir, f"inpaint_{flag_name}.pt")

    result_this_flag = torch.load(save_path, map_location="cpu")

    flags = result_this_flag["flags"].to(device)

    print("Loaded:", save_path)
    print(f"Flag {flag_id}: {flag_name}")
    print("flags shape:", flags.shape)

    # --------------------------------------------------
    # 2.1 Coherent time average for original and flagged data
    # --------------------------------------------------
    avg_times = ba.utils.split_into_groups(
        torch.arange(Ntimes),
        Nelem=50,
    )

    # Original data
    vd_avg = vd.time_average(
        avg_times,
        rephase=True,
        inplace=False,
    )

    # Original data with flagged pixels removed
    vd_flg = vd * ~flags

    vd_flg_avg = vd_flg.time_average(
        avg_times,
        wgts=1 / (flags * 1e5 + 1e0),
        rephase=True,
        inplace=False,
    )

    # ==================================================
    # 3. Fractional absolute delay residual:
    #
    #       ||V_data| - |V_inpainted|| / |V_data|
    #
    #    plot range: -1500 < tau < 1500 ns
    #    shaded region: 800 < |tau| < 1500 ns
    # ==================================================

    fig, ax = plt.subplots(1, 1, figsize=(10, 5))

    # --------------------------------------------------
    # 3.1 Compute original data delay spectrum
    # --------------------------------------------------
    data_delay = FT1(vd_avg.data[0, 0].mean(0))[:].abs().mean(0)

    # Plot range: -1500 < tau < 1500 ns
    plot_mask = (FT1.freqs > -1500) & (FT1.freqs < 1500)

    tau_plot = FT1.freqs[plot_mask]
    data_delay_plot = data_delay[plot_mask]

    # --------------------------------------------------
    # 3.2 Loop over inpainting algorithms and plot residual ratio
    # --------------------------------------------------
    for Algorithm_id, Algorithm_name in enumerate(Algorithm_names, start=1):

        inp_y = result_this_flag["inpaint"][Algorithm_name]["inp_y"].to(device)

        vd_inp = vd.copy(copydata=True)
        vd_inp.data = inp_y[None, None]

        vd_inp_avg = vd_inp.time_average(
            avg_times,
            rephase=True,
            inplace=False,
        )

        inp_delay = FT1(vd_inp_avg.data[0, 0].mean(0))[:].abs().mean(0)
        inp_delay_plot = inp_delay[plot_mask]

        # Fractional absolute residual:
        # ||V_data| - |V_inpainted|| / |V_data|
        frac_residual = (data_delay_plot - inp_delay_plot).abs() / (
            data_delay_plot + 1e-30
        )

        style_id = Algorithm_id - 1

        ax.plot(
            tau_plot.detach().cpu(),
            frac_residual.detach().cpu(),
            linestyle=linestyles[style_id],
            marker=markers[style_id],
            markevery=markevery,
            markersize=4,
            linewidth=linewidths[style_id],
            alpha=0.9,
            label=f"Algorithm {Algorithm_id}: {Algorithm_name}",
        )

    # --------------------------------------------------
    # 3.3 Shaded delay regions: 800 < |tau| < 1500 ns
    # --------------------------------------------------
    ax.axvspan(
        -1500,
        -800,
        alpha=0.2,
        label=r"$800 < |\tau| < 1500$ ns",
    )

    ax.axvspan(
        800,
        1500,
        alpha=0.2,
    )

    # --------------------------------------------------
    # 3.4 Plot formatting
    # --------------------------------------------------
    ax.axhline(
        0.0,
        color="k",
        ls=":",
        lw=1,
        label="residual = 0",
    )

    ax.grid()
    ax.set_xlim(-1500, 1500)

    ax.set_xlabel("Delay [ns]")
    ax.set_ylabel(
        r"$\frac{\left||V_{\rm data}(\tau)| - |V_{\rm inpainted}(\tau)|\right|}"
        r"{|V_{\rm data}(\tau)|}$",
        fontsize=16,
    )

    add_flag_title(
        ax,
        flag_id,
        flag_name,
        subtitle=(
            r"Fractional Delay Residual: "
            r"$\left||V_{\rm data}| - |V_{\rm inpainted}|\right|/|V_{\rm data}|$"
        ),
        y=1.45,
    )

    ax.legend(
        loc="lower center",
        bbox_to_anchor=(0.5, 1.08),
        ncol=3,
        frameon=False,
        fontsize=9,
    )

    plt.tight_layout()
    plt.show()
```
