# 1. Averaged visibility as a function of frequency and fractional residual

### 1. For `["real"]` and `["fullband_short_time"]`, some frequency slices become entirely zero after inpainting. As a result, information is lost when averaging over these frequencies, leading to a large fractional residual compared with the true visibility.


```python
import os
import torch
import matplotlib.pyplot as plt
import pandas as pd


# ==================================================
# 0. Helper: title style
# ==================================================

def add_flag_title(ax, flag_id, flag_name, subtitle, y=1.32):
    ax.text(
        0.5,
        y,
        f"Flag {flag_id}: {flag_name}",
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        color="darkred",
        fontsize=12,
        fontweight="bold",
    )

    ax.text(
        0.5,
        y - 0.12,
        subtitle,
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        color="black",
        fontsize=12,
    )


def shade_flagged_freqs(ax, flags, freqs, label="Flagged frequency region"):
    """
    Shade frequency channels that are flagged at least once
    over time/baseline.

    flags shape can be:
        (1, Nbls, Nfreqs)
        (Ntimes, Nbls, Nfreqs)
    """

    # flags: (..., Nfreqs) -> (Nfreqs,)
    flag_freq = flags.any(dim=tuple(range(flags.ndim - 1))).detach().cpu()

    freqs_cpu = freqs.detach().cpu() if torch.is_tensor(freqs) else torch.tensor(freqs)

    df = freqs_cpu[1] - freqs_cpu[0]

    in_region = False
    start = None
    first_label = True

    for i, is_flag in enumerate(flag_freq):

        if is_flag and not in_region:
            start = i
            in_region = True

        if in_region and ((not is_flag) or (i == len(flag_freq) - 1)):
            end = i if not is_flag else i + 1

            x0 = freqs_cpu[start] - df / 2
            x1 = freqs_cpu[end - 1] + df / 2

            ax.axvspan(
                x0,
                x1,
                alpha=0.18,
                color="gray",
                label=label if first_label else None,
            )

            first_label = False
            in_region = False


# ==================================================
# 1. Algorithms to compare
# ==================================================

Algorithm_names = [
    "freq_1d",
    "time_1d_freq_1d",
    "time_freq_2d",
    "time_freq_2d_freq_1d",
]

flag_types_all = [
    ["real"],
    ["narrowband_full_time"],
    ["wideband_full_time"],
    ["ultraband_short_time"],
    ["fullband_short_time"],
    ["narrowband_short_time"],
    ["narrowband_intermittent"],
]


# ==================================================
# 1.5 Store RMS results
# ==================================================

rms_records = []


# ==================================================
# 2. Loop over flag cases
# ==================================================

for flag_id, flag_types in enumerate(flag_types_all, start=1):

    flag_name = "+".join(flag_types)
    save_path = os.path.join(save_dir, f"inpaint_{flag_name}.pt")

    result_this_flag = torch.load(save_path, map_location="cpu")

    flags = result_this_flag["flags"].to(device).bool()

    print("Loaded:", save_path)
    print(f"Flag {flag_id}: {flag_name}")
    print("flags shape:", flags.shape)

    # --------------------------------------------------
    # 2.1 Coherent time average for original data
    # --------------------------------------------------

    avg_times = ba.utils.split_into_groups(
        torch.arange(Ntimes),
        Nelem=50,
    )

    vd_avg = vd.time_average(
        avg_times,
        rephase=True,
        inplace=False,
    )

    # Averaged original data for V(nu) plot
    y_true = vd_avg.data[0, 0].mean(0).abs().mean(0)

    # Original unaveraged data for RMS
    data_y = vd.data[0, 0].to(device)      # expected shape: (Ntimes, Nbls, Nfreqs)

    # ==================================================
    # 3. V(frequency): compare different algorithms
    # ==================================================

    fig, ax = plt.subplots(1, 1, figsize=(10, 5))

    shade_flagged_freqs(ax, flags, freqs)

    ax.plot(
        freqs,
        y_true,
        label="Original data",
    )

    y_inp_dict = {}
    rms_dict = {}

    for Algorithm_id, Algorithm_name in enumerate(Algorithm_names, start=1):

        inp_y = result_this_flag["inpaint"][Algorithm_name]["inp_y"].to(device)
        # expected shape: (Ntimes, Nbls, Nfreqs)

        # ==================================================
        # RMS on original, unaveraged flagged pixels
        # Use broadcasting, same logic as: vd_flg = vd * ~flags
        # ==================================================

        diff = inp_y - data_y

        # flags may be shape (1, Nbls, Nfreqs)
        # diff is shape (Ntimes, Nbls, Nfreqs)
        # expand flags to the full unaveraged data shape
        flag_w = flags.to(diff.real.dtype).expand_as(diff.real)

        if flag_w.sum() > 0:
            rms_flag = torch.sqrt(
                ((diff.abs() ** 2) * flag_w).sum() / flag_w.sum()
            ).item()
        else:
            rms_flag = float("nan")

        rms_dict[Algorithm_name] = rms_flag

        rms_records.append(
            {
                "flag_id": flag_id,
                "flag_name": flag_name,
                "algorithm_id": Algorithm_id,
                "algorithm_name": Algorithm_name,
                "rms_flagged_pixels_unaveraged": rms_flag,
            }
        )

        # ==================================================
        # Time average only for V(nu) plot
        # ==================================================

        vd_inp = vd.copy(copydata=True)
        vd_inp.data = inp_y[None, None]

        x = vd_inp.data[0, 0]   # shape: (Ntimes, Nbls, Nfreqs)

        # 如果某个 [time, baseline, :] 的频率全是 0，
        # 则这一整条不参与 time_average
        nonzero_vec = x.abs().sum(dim=-1) > 0       # shape: (Ntimes, Nbls)

        wgts = nonzero_vec[..., None].to(x.real.dtype).expand_as(x.real)

        vd_inp_avg = vd_inp.time_average(
            avg_times,
            wgts=wgts[None, None],
            rephase=True,
            inplace=False,
        )

        y_inp = vd_inp_avg.data[0, 0].mean(0).abs().mean(0)

        y_inp_dict[Algorithm_name] = y_inp

        ax.plot(
            freqs,
            y_inp,
            ls="--",
            label=f"Algorithm {Algorithm_id}: {Algorithm_name}",
        )

    ax.set_yscale("log")
    ax.set_xlabel("Frequency [MHz]")
    ax.set_ylabel(r"$|V(\nu)|$")

    add_flag_title(
        ax,
        flag_id,
        flag_name,
        subtitle=r"$V(\nu)$",
        y=1.45,
    )

    ax.legend(
        loc="lower center",
        bbox_to_anchor=(0.5, 1.0),
        ncol=3,
        frameon=False,
        fontsize=10,
    )

    plt.tight_layout()
    plt.show()

    # ==================================================
    # 4. Fractional residual in V(frequency)
    #    Use the exact same y_inp plotted in Section 3
    # ==================================================

    fig, ax = plt.subplots(1, 1, figsize=(10, 5))

    shade_flagged_freqs(ax, flags, freqs)

    for Algorithm_id, Algorithm_name in enumerate(Algorithm_names, start=1):

        y_inp = y_inp_dict[Algorithm_name]

        frac_res = (y_inp - y_true).abs() / y_true.abs().clamp_min(1e-30)

        rms_flag = rms_dict[Algorithm_name]

        ax.plot(
            freqs,
            frac_res,
            ls="--",
            label=(
                f"Algorithm {Algorithm_id}: {Algorithm_name} "
                #f"(RMS={rms_flag:.3e})"
            ),
        )

    ax.set_xlabel("Frequency [MHz]")
    ax.set_ylabel(
        r"$\left||V_{\rm data}(\nu)| - |V_{\rm inpainted}(\nu)|\right|"
        r"/|V_{\rm data}(\nu)|$"
    )

    add_flag_title(
        ax,
        flag_id,
        flag_name,
        subtitle=r"Fractional residual in $V(\nu)$",
        y=1.45,
    )

    ax.legend(
        loc="lower center",
        bbox_to_anchor=(0.5, 1.0),
        ncol=2,
        frameon=False,
        fontsize=10,
    )

    plt.tight_layout()
    plt.show()


```

```python
import os
import torch
import matplotlib.pyplot as plt
import pandas as pd


# ==================================================
# 0. Helper: title style
# ==================================================

def add_flag_title(ax, flag_id, flag_name, subtitle, y=1.32):
    ax.text(
        0.5,
        y,
        f"Flag {flag_id}: {flag_name}",
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        color="darkred",
        fontsize=12,
        fontweight="bold",
    )

    ax.text(
        0.5,
        y - 0.12,
        subtitle,
        transform=ax.transAxes,
        ha="center",
        va="bottom",
        color="black",
        fontsize=12,
    )


def shade_flagged_freqs(ax, flags, freqs, label="Flagged frequency region"):
    """
    Shade frequency channels that are flagged at least once
    over time/baseline.

    flags shape can be:
        (1, Nbls, Nfreqs)
        (Ntimes, Nbls, Nfreqs)
    """

    flag_freq = flags.any(dim=tuple(range(flags.ndim - 1))).detach().cpu()

    freqs_cpu = freqs.detach().cpu() if torch.is_tensor(freqs) else torch.tensor(freqs)

    df = freqs_cpu[1] - freqs_cpu[0]

    in_region = False
    start = None
    first_label = True

    for i, is_flag in enumerate(flag_freq):

        is_flag = bool(is_flag)

        if is_flag and not in_region:
            start = i
            in_region = True

        if in_region and ((not is_flag) or (i == len(flag_freq) - 1)):
            end = i if not is_flag else i + 1

            x0 = (freqs_cpu[start] - df / 2).item()
            x1 = (freqs_cpu[end - 1] + df / 2).item()

            ax.axvspan(
                x0,
                x1,
                alpha=0.18,
                color="gray",
                label=label if first_label else None,
            )

            first_label = False
            in_region = False


# ==================================================
# 1. Algorithms to compare
# ==================================================

Algorithm_names = [
    "freq_1d",
    "time_1d_freq_1d",
    "time_freq_2d",
    "time_freq_2d_freq_1d",
]

flag_types_all = [
    ["real"],
    ["narrowband_full_time"],
    ["wideband_full_time"],
    ["ultraband_short_time"],
    ["fullband_short_time"],
    ["narrowband_short_time"],
    ["narrowband_intermittent"],
]


# ==================================================
# 1.5 Store RMS results
# ==================================================

rms_records = []


# ==================================================
# 2. Loop over flag cases
# ==================================================

for flag_id, flag_types in enumerate(flag_types_all, start=1):

    flag_name = "+".join(flag_types)
    save_path = os.path.join(save_dir, f"inpaint_{flag_name}.pt")

    result_this_flag = torch.load(save_path, map_location="cpu")

    flags = result_this_flag["flags"].to(device).bool()

    print("Loaded:", save_path)
    print(f"Flag {flag_id}: {flag_name}")
    print("flags shape:", flags.shape)

    # --------------------------------------------------
    # 2.1 Coherent time average for original data
    # --------------------------------------------------

    avg_times = ba.utils.split_into_groups(
        torch.arange(Ntimes),
        Nelem=50,
    )

    vd_avg = vd.time_average(
        avg_times,
        rephase=True,
        inplace=False,
    )

    y_true = vd_avg.data[0, 0].mean(0).abs().mean(0)

    data_y = vd.data[0, 0].to(device)

    # ==================================================
    # 3. V(frequency): compare different algorithms
    # ==================================================

    fig, ax = plt.subplots(1, 1, figsize=(10, 5))

    shade_flagged_freqs(ax, flags, freqs)

    ax.plot(
        freqs,
        y_true,
        label="Original data",
    )

    y_inp_dict = {}
    rms_dict = {}

    for Algorithm_id, Algorithm_name in enumerate(Algorithm_names, start=1):

        inp_y = result_this_flag["inpaint"][Algorithm_name]["inp_y"].to(device)

        # ==================================================
        # RMS on original, unaveraged flagged pixels
        # ==================================================

        diff = inp_y - data_y

        flag_w = flags.to(diff.real.dtype).expand_as(diff.real)

        if flag_w.sum().item() > 0:
            rms_flag = torch.sqrt(
                ((diff.abs() ** 2) * flag_w).sum() / flag_w.sum()
            ).item()
        else:
            rms_flag = float("nan")

        rms_dict[Algorithm_name] = rms_flag

        rms_records.append(
            {
                "flag_id": flag_id,
                "flag_name": flag_name,
                "algorithm_id": Algorithm_id,
                "algorithm_name": Algorithm_name,
                "rms_flagged_pixels_unaveraged": rms_flag,
            }
        )

        # ==================================================
        # Time average only for V(nu) plot
        # ==================================================

        vd_inp = vd.copy(copydata=True)
        vd_inp.data = inp_y[None, None]

        x = vd_inp.data[0, 0]

        nonzero_vec = x.abs().sum(dim=-1) > 0
        wgts = nonzero_vec[..., None].to(x.real.dtype).expand_as(x.real)

        vd_inp_avg = vd_inp.time_average(
            avg_times,
            wgts=wgts[None, None],
            rephase=True,
            inplace=False,
        )

        y_inp = vd_inp_avg.data[0, 0].mean(0).abs().mean(0)

        y_inp_dict[Algorithm_name] = y_inp

        ax.plot(
            freqs,
            y_inp,
            ls="--",
            label=f"Algorithm {Algorithm_id}: {Algorithm_name}",
        )

    ax.set_yscale("log")
    ax.set_xlabel("Frequency [MHz]")
    ax.set_ylabel(r"$|V(\nu)|$")

    add_flag_title(
        ax,
        flag_id,
        flag_name,
        subtitle=r"$V(\nu)$",
        y=1.45,
    )

    ax.legend(
        loc="lower center",
        bbox_to_anchor=(0.5, 1.0),
        ncol=3,
        frameon=False,
        fontsize=10,
    )

    plt.tight_layout()
    plt.show()

    # ==================================================
    # 4. Fractional residual in V(frequency)
    # ==================================================

    fig, ax = plt.subplots(1, 1, figsize=(10, 5))

    shade_flagged_freqs(ax, flags, freqs)

    for Algorithm_id, Algorithm_name in enumerate(Algorithm_names, start=1):

        # 只跳过 real 和 fullband_short_time 里的 freq_1d
        if Algorithm_name == "freq_1d" and flag_types in [["real"], ["fullband_short_time"]]:
            continue

        y_inp = y_inp_dict[Algorithm_name]

        frac_res = (y_inp - y_true).abs() / y_true.abs().clamp_min(1e-30)

        ax.plot(
            freqs,
            frac_res,
            ls="--",
            label=f"Algorithm {Algorithm_id}: {Algorithm_name}",
        )

    ax.set_xlabel("Frequency [MHz]")
    ax.set_ylabel(
        r"$\left||V_{\rm data}(\nu)| - |V_{\rm inpainted}(\nu)|\right|"
        r"/|V_{\rm data}(\nu)|$"
    )

    add_flag_title(
        ax,
        flag_id,
        flag_name,
        subtitle=r"Fractional residual in $V(\nu)$",
        y=1.45,
    )

    ax.legend(
        loc="lower center",
        bbox_to_anchor=(0.5, 1.0),
        ncol=2,
        frameon=False,
        fontsize=10,
    )

    plt.tight_layout()
    plt.show()
```
