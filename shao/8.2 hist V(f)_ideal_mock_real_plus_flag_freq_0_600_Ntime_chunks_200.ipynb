## Results for More Complex Flagging Patterns

For the more complex flagging cases, where each simulated flagging pattern is combined with real flags:

1. The `freq` method fails to inpaint the flagged data reliably, producing an RMS approximately two orders of magnitude larger than that of the `2d_1d` method.


2. For baseline `(0, 1)`, the `2d_1d` method achieves an RMS approximately one order of magnitude lower than that of `1d_1d` for the `wideband_full_time` and `narrowband_long_time` flagging patterns.


3. For baselines `(0, 5)` and `(0, 10)`, the `2d_1d` method achieves an RMS approximately one order of magnitude lower than that of `1d_1d` across all tested flagging patterns.

4. Compared with the simpler flagging patterns, the advantage of `2d_1d` over `1d_1d` and `freq` becomes more pronounced for the more complex flagging patterns, particularly for `narrowband_long_time`.



```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np
import pandas as pd
import torch
from pathlib import Path

import sys
sys.path.append("./")

from utils_load import load_hist_residual


# ============================================================
# Settings
# ============================================================

torch.set_default_dtype(torch.float64)

time_ints = range(80, 1680)
polsi = "ee"
add_real_flags = True

blall = [(0, 1), (0, 5), (0, 10)]

inpaint_method_all = ["2d_1d", "1d_1d", "freq"]

flag_types_all = [
    None,
    "narrowband_full_time",
    "wideband_full_time",
    "ultraband_short_time",
    "fullband_short_time",
    "narrowband_short_time",
    "narrowband_long_time",
    "wideband_long_time",
    "randn",
]


# ============================================================
# Frequency and time-chunk configurations
# ============================================================

configurations = [
    {"label": "freq_0_600, chunks_50", "freq_chans": range(0, 600), "Ntime_chunks": 50},
    {"label": "freq_0_600, chunks_100", "freq_chans": range(0, 600), "Ntime_chunks": 100},
    {"label": "freq_0_600, chunks_200", "freq_chans": range(0, 600), "Ntime_chunks": 200},
    {"label": "freq_0_600, chunks_400", "freq_chans": range(0, 600), "Ntime_chunks": 400},
    {"label": "freq_0_600, chunks_800", "freq_chans": range(0, 600), "Ntime_chunks": 800},
    {"label": "freq_150_500, chunks_200", "freq_chans": range(150, 500), "Ntime_chunks": 200},
    {"label": "freq_200_450, chunks_200", "freq_chans": range(200, 450), "Ntime_chunks": 200},
    {"label": "freq_250_400, chunks_200", "freq_chans": range(250, 400), "Ntime_chunks": 200},
]


# ============================================================
# Plot settings
# ============================================================

root_dir = "hist_residual"

hist_bins = 80
hist_quantile = 0.995
display_fraction_x = 0.8

save_figures = True
save_tables = True

figure_dir = Path("hist_comparison_figures")
table_dir = Path("hist_comparison_tables")

figure_dir.mkdir(parents=True, exist_ok=True)
table_dir.mkdir(parents=True, exist_ok=True)

# Line styles for histogram configurations
hist_linestyles = [
    "-",
    "--",
    "-.",
    ":",
    (0, (5, 1)),
    (0, (3, 1, 1, 1)),
    (0, (1, 1)),
    (0, (5, 2, 1, 2)),
]

# Line styles and markers for inpainting methods
method_linestyles = ["-", "--", "-."]
method_markers = ["o", "s", "^"]


# ============================================================
# Helper functions
# ============================================================

def get_flag_title(flag_typesi, add_real_flags):
    if flag_typesi is None:
        return "Real Flags" if add_real_flags else "No Flags"

    title = flag_typesi.replace("_", " ").title()
    return f"Real Flags + {title}" if add_real_flags else title


def clean_hist_values(hist_residual):
    values = hist_residual.detach().cpu().reshape(-1).numpy()
    return values[np.isfinite(values)]


def calculate_statistics(values):
    if values.size == 0:
        return np.nan, np.nan, np.nan, 0

    mean = np.mean(values)
    std = np.std(values, ddof=1) if values.size > 1 else 0.0
    rmse = np.sqrt(np.mean(values**2))

    return mean, std, rmse, values.size


# ============================================================
# Main loop
# ============================================================

all_statistics = []

for bl in blall:

    nrows = len(flag_types_all)
    ncols = len(inpaint_method_all) + 2

    fig, axes = plt.subplots(nrows, ncols, figsize=(4.2 * ncols, 3.1 * nrows), squeeze=False)

    baseline_statistics = []

    for row_idx, flag_typesi in enumerate(flag_types_all):

        flag_title = get_flag_title(flag_typesi, add_real_flags)
        flag_results = {}

        # ========================================================
        # Load all histogram results for this flag
        # ========================================================

        for inpaint_methodi in inpaint_method_all:

            flag_results[inpaint_methodi] = []

            for config_idx, config in enumerate(configurations):

                hist_residual = load_hist_residual(
                    bl=bl, flag_typesi=flag_typesi, add_real_flags=add_real_flags,
                    inpaint_method=inpaint_methodi, freq_chans=config["freq_chans"],
                    time_ints=time_ints, polsi=polsi, Ntime_chunks=config["Ntime_chunks"],
                    root_dir=root_dir, device="cpu",
                )

                values = clean_hist_values(hist_residual)
                mean, std, rmse, sample_number = calculate_statistics(values)

                result = {
                    "config_idx": config_idx,
                    "label": config["label"],
                    "freq_start": config["freq_chans"].start,
                    "freq_stop": config["freq_chans"].stop,
                    "Ntime_chunks": config["Ntime_chunks"],
                    "values": values,
                    "mean": mean,
                    "std": std,
                    "rmse": rmse,
                    "N": sample_number,
                }

                flag_results[inpaint_methodi].append(result)

                baseline_statistics.append({
                    "baseline": str(bl),
                    "polarization": polsi,
                    "flag": flag_typesi,
                    "method": inpaint_methodi,
                    "config_number": config_idx + 1,
                    "configuration": config["label"],
                    "freq_start": config["freq_chans"].start,
                    "freq_stop": config["freq_chans"].stop,
                    "Ntime_chunks": config["Ntime_chunks"],
                    "N": sample_number,
                    "mean": mean,
                    "std": std,
                    "rmse": rmse,
                })

        # ========================================================
        # Column 1: RMSE versus Ntime_chunks
        # Only use frequency channels 0–600
        # ========================================================

        ax_chunk = axes[row_idx, 0]

        for method_idx, inpaint_methodi in enumerate(inpaint_method_all):

            chunk_results = [
                result for result in flag_results[inpaint_methodi]
                if result["freq_start"] == 0 and result["freq_stop"] == 600
            ]

            chunk_results = sorted(chunk_results, key=lambda result: result["Ntime_chunks"])

            chunk_x = [result["Ntime_chunks"] for result in chunk_results]
            chunk_rmse = [result["rmse"] for result in chunk_results]

            ax_chunk.plot(
                chunk_x, chunk_rmse,
                marker=method_markers[method_idx % len(method_markers)],
                linestyle=method_linestyles[method_idx % len(method_linestyles)],
                linewidth=1.8, markersize=6, markerfacecolor="none",
                markeredgewidth=1.2, label=inpaint_methodi, zorder=method_idx + 2,
            )

        ax_chunk.set_xscale("log", base=2)
        ax_chunk.set_yscale("log")

        ax_chunk.set_xticks([50, 100, 200, 400, 800])
        ax_chunk.set_xticklabels(["50", "100", "200", "400", "800"])

        ax_chunk.set_xlabel("Ntime chunks", fontsize=10)
        ax_chunk.set_ylabel(f"{flag_title}\nResidual RMSE", fontsize=10)

        ax_chunk.tick_params(axis="both", labelsize=8)
        ax_chunk.grid(True, which="both", alpha=0.25)

        if row_idx == 0:
            ax_chunk.set_title("RMSE vs Chunks\nFreq 0–600", fontsize=12, fontweight="bold")
            ax_chunk.legend(fontsize=8, loc="best", handlelength=2.5)

        # ========================================================
        # Column 2: RMSE versus frequency range
        # Only use Ntime_chunks = 200
        # ========================================================

        ax_freq = axes[row_idx, 1]

        freq_configs = [(0, 600), (150, 500), (200, 450), (250, 400)]

        freq_positions = np.arange(len(freq_configs))
        freq_labels = [f"{start}–{stop}" for start, stop in freq_configs]

        for method_idx, inpaint_methodi in enumerate(inpaint_method_all):

            freq_results = []

            for freq_start, freq_stop in freq_configs:

                matching_results = [
                    result for result in flag_results[inpaint_methodi]
                    if result["freq_start"] == freq_start
                    and result["freq_stop"] == freq_stop
                    and result["Ntime_chunks"] == 200
                ]

                freq_results.append(matching_results[0] if matching_results else None)

            freq_rmse = [
                result["rmse"] if result is not None else np.nan
                for result in freq_results
            ]

            ax_freq.plot(
                freq_positions, freq_rmse,
                marker=method_markers[method_idx % len(method_markers)],
                linestyle=method_linestyles[method_idx % len(method_linestyles)],
                linewidth=1.8, markersize=6, markerfacecolor="none",
                markeredgewidth=1.2, label=inpaint_methodi, zorder=method_idx + 2,
            )

        ax_freq.set_yscale("log")

        ax_freq.set_xticks(freq_positions)
        ax_freq.set_xticklabels(freq_labels, rotation=30, ha="right")

        ax_freq.set_xlabel("Frequency channels", fontsize=10)
        ax_freq.set_ylabel("Residual RMSE", fontsize=10)

        ax_freq.tick_params(axis="both", labelsize=8)
        ax_freq.grid(True, which="both", alpha=0.25)

        if row_idx == 0:
            ax_freq.set_title("RMSE vs Frequency\nChunks 200", fontsize=12, fontweight="bold")
            ax_freq.legend(fontsize=8, loc="best", handlelength=2.5)

        # ========================================================
        # Remaining columns: histogram for each algorithm
        # ========================================================

        for method_idx, inpaint_methodi in enumerate(inpaint_method_all):

            ax = axes[row_idx, method_idx + 2]
            subplot_results = flag_results[inpaint_methodi]

            valid_arrays = [
                result["values"]
                for result in subplot_results
                if result["values"].size > 0
            ]

            if not valid_arrays:
                ax.text(0.5, 0.5, "No data", ha="center", va="center", transform=ax.transAxes)
                continue

            hist_all = np.concatenate(valid_arrays)
            hist_limit = np.quantile(np.abs(hist_all), hist_quantile)

            if not np.isfinite(hist_limit) or hist_limit <= 0:
                hist_limit = np.max(np.abs(hist_all))

            if not np.isfinite(hist_limit) or hist_limit <= 0:
                hist_limit = 1.0

            x_limit = hist_limit * display_fraction_x

            for result in subplot_results:

                if result["values"].size == 0:
                    continue

                config_idx = result["config_idx"]

                ax.hist(
                    result["values"], bins=hist_bins, range=(-hist_limit, hist_limit),
                    density=True, histtype="step", linewidth=1.8,
                    linestyle=hist_linestyles[config_idx % len(hist_linestyles)],
                    label=str(config_idx + 1), zorder=config_idx + 1,
                )

            ax.axvline(0, color="k", linestyle="--", linewidth=0.8)

            ax.set_xlim(-x_limit, x_limit)
            ax.set_yscale("log")

            ax.tick_params(axis="both", labelsize=8)

            ax.legend(
                fontsize=6, loc="upper right", frameon=True, handlelength=2.0,
                labelspacing=0.2, borderpad=0.3, ncol=2,
            )

            if row_idx == nrows - 1:
                ax.set_xlabel(
                    r"$(|V_{\mathrm{inp}}|-|V_{\mathrm{true}}|)/|V_{\mathrm{true}}|$",
                    fontsize=10,
                )

            if method_idx == 0:
                ax.set_ylabel("Density", fontsize=10)

            if row_idx == 0:
                ax.set_title(inpaint_methodi, fontsize=13, fontweight="bold")

    # ========================================================
    # Configuration legend outside all subplots
    # ========================================================

    configuration_handles = [
        Line2D(
            [0], [0], color="black", linewidth=1.8,
            linestyle=hist_linestyles[i % len(hist_linestyles)],
            label=f"{i + 1}: {config['label']}",
        )
        for i, config in enumerate(configurations)
    ]

    fig.legend(
        handles=configuration_handles, loc="upper center",
        bbox_to_anchor=(0.5, 0.972), ncol=4, fontsize=8,
        frameon=True, handlelength=2.5, columnspacing=1.5,
    )

    # ========================================================
    # Figure formatting
    # ========================================================

    fig.suptitle(f"Baseline {bl}, Pol {polsi}", fontsize=16, y=0.998)

    fig.subplots_adjust(
        top=0.90, bottom=0.06, left=0.07, right=0.99,
        hspace=0.40, wspace=0.32,
    )

    if save_figures:
        figure_name = f"hist_rmse_comparison_bl_{bl[0]}_{bl[1]}_pol_{polsi}.png"
        fig.savefig(figure_dir / figure_name, dpi=200, bbox_inches="tight")

    plt.show()
    plt.close(fig)

    # ========================================================
    # Statistics table
    # Smaller RMSE means better
    # ========================================================

    baseline_df = pd.DataFrame(baseline_statistics)

    baseline_df["rank"] = baseline_df.groupby(["flag", "method"])["rmse"].rank(method="min")

    display_columns = [
        "baseline",
        "flag",
        "method",
        "config_number",
        "configuration",
        "freq_start",
        "freq_stop",
        "Ntime_chunks",
        "N",
        "mean",
        "std",
        "rmse",
        "rank",
    ]

    display(
        baseline_df[display_columns]
        .sort_values(["flag", "method", "rank"])
        .reset_index(drop=True)
    )

    # ========================================================
    # Best configuration for each flag and algorithm
    # ========================================================

    best_configuration_df = (
        baseline_df
        .sort_values("rmse")
        .groupby(["flag", "method"], as_index=False)
        .first()
    )

    print(f"\nBest configurations for baseline {bl}:")

    display(
        best_configuration_df[
            [
                "flag",
                "method",
                "config_number",
                "configuration",
                "mean",
                "std",
                "rmse",
            ]
        ]
    )

    if save_tables:

        baseline_df.to_csv(
            table_dir / f"hist_rmse_statistics_bl_{bl[0]}_{bl[1]}_pol_{polsi}.csv",
            index=False,
        )

        best_configuration_df.to_csv(
            table_dir / f"best_rmse_configurations_bl_{bl[0]}_{bl[1]}_pol_{polsi}.csv",
            index=False,
        )

    all_statistics.append(baseline_df)


# ============================================================
# Combined table for all baselines
# ============================================================

if all_statistics:

    all_statistics_df = pd.concat(all_statistics, ignore_index=True)

    all_statistics_df["overall_rank"] = all_statistics_df.groupby(
        ["baseline", "flag", "method"]
    )["rmse"].rank(method="min")

    display(
        all_statistics_df[
            [
                "baseline",
                "flag",
                "method",
                "config_number",
                "configuration",
                "freq_start",
                "freq_stop",
                "Ntime_chunks",
                "N",
                "mean",
                "std",
                "rmse",
                "overall_rank",
            ]
        ]
        .sort_values(["baseline", "flag", "method", "overall_rank"])
        .reset_index(drop=True)
    )

    if save_tables:
        all_statistics_df.to_csv(
            table_dir / f"hist_rmse_statistics_all_baselines_pol_{polsi}.csv",
            index=False,
        )
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np
import pandas as pd
import torch
from pathlib import Path

import sys
sys.path.append("./")

from utils_load import load_hist_residual


# ============================================================
# Settings
# ============================================================

torch.set_default_dtype(torch.float64)

time_ints = range(80, 1680)
polsi = "ee"
add_real_flags = True

blall = [(0, 1), (0, 5), (0, 10)]

inpaint_method_all = ["2d_1d", "1d_1d", "freq"]

flag_types_all = [
    None,
    "narrowband_full_time",
    "wideband_full_time",
    "ultraband_short_time",
    "fullband_short_time",
    "narrowband_short_time",
    "narrowband_long_time",
    "wideband_long_time",
    "randn",
]


# ============================================================
# Frequency and time-chunk configurations
# ============================================================

configurations = [
    {"label": "freq_0_600, chunks_50", "freq_chans": range(0, 600), "Ntime_chunks": 50},
    {"label": "freq_0_600, chunks_100", "freq_chans": range(0, 600), "Ntime_chunks": 100},
    {"label": "freq_0_600, chunks_200", "freq_chans": range(0, 600), "Ntime_chunks": 200},
    {"label": "freq_0_600, chunks_400", "freq_chans": range(0, 600), "Ntime_chunks": 400},
    {"label": "freq_0_600, chunks_800", "freq_chans": range(0, 600), "Ntime_chunks": 800},
    {"label": "freq_150_500, chunks_200", "freq_chans": range(150, 500), "Ntime_chunks": 200},
    {"label": "freq_200_450, chunks_200", "freq_chans": range(200, 450), "Ntime_chunks": 200},
    {"label": "freq_250_400, chunks_200", "freq_chans": range(250, 400), "Ntime_chunks": 200},
]


# ============================================================
# Plot settings
# ============================================================

root_dir = "hist_residual"

hist_bins = 80
hist_quantile = 0.995
display_fraction_x = 0.8

save_figures = True
save_tables = True

figure_dir = Path("hist_comparison_figures")
table_dir = Path("hist_comparison_tables")

figure_dir.mkdir(parents=True, exist_ok=True)
table_dir.mkdir(parents=True, exist_ok=True)

linestyles = [
    "-",
    "--",
    "-.",
    ":",
    (0, (5, 1)),
    (0, (3, 1, 1, 1)),
    (0, (1, 1)),
    (0, (5, 2, 1, 2)),
]


# ============================================================
# Helper functions
# ============================================================

def get_flag_title(flag_typesi, add_real_flags):
    if flag_typesi is None:
        return "Real Flags" if add_real_flags else "No Flags"

    title = flag_typesi.replace("_", " ").title()
    return f"Real Flags + {title}" if add_real_flags else title


def clean_hist_values(hist_residual):
    values = hist_residual.detach().cpu().reshape(-1).numpy()
    return values[np.isfinite(values)]


def calculate_statistics(values):
    if values.size == 0:
        return np.nan, np.nan, np.nan, 0

    mean = np.mean(values)
    std = np.std(values, ddof=1) if values.size > 1 else 0.0
    rmse = np.sqrt(np.mean(values**2))

    return mean, std, rmse, values.size


# ============================================================
# Main loop
# ============================================================

all_statistics = []

for bl in blall:

    # Each flag uses three rows:
    # row 1: histogram
    # row 2: RMSE versus chunks
    # row 3: RMSE versus frequency
    nrows = len(flag_types_all) * 3
    ncols = len(inpaint_method_all)

    fig, axes = plt.subplots(
        nrows,
        ncols,
        figsize=(4.8 * ncols, 2.8 * nrows),
        squeeze=False,
    )

    baseline_statistics = []

    for flag_idx, flag_typesi in enumerate(flag_types_all):

        flag_title = get_flag_title(flag_typesi, add_real_flags)
        flag_results = {}

        hist_row = flag_idx * 3
        chunk_row = hist_row + 1
        freq_row = hist_row + 2

        # ========================================================
        # Load all configurations for this flag
        # ========================================================

        for inpaint_methodi in inpaint_method_all:

            flag_results[inpaint_methodi] = []

            for config_idx, config in enumerate(configurations):

                hist_residual = load_hist_residual(
                    bl=bl, flag_typesi=flag_typesi, add_real_flags=add_real_flags,
                    inpaint_method=inpaint_methodi, freq_chans=config["freq_chans"],
                    time_ints=time_ints, polsi=polsi, Ntime_chunks=config["Ntime_chunks"],
                    root_dir=root_dir, device="cpu",
                )

                values = clean_hist_values(hist_residual)
                mean, std, rmse, sample_number = calculate_statistics(values)

                result = {
                    "config_idx": config_idx,
                    "label": config["label"],
                    "freq_start": config["freq_chans"].start,
                    "freq_stop": config["freq_chans"].stop,
                    "Ntime_chunks": config["Ntime_chunks"],
                    "values": values,
                    "mean": mean,
                    "std": std,
                    "rmse": rmse,
                    "N": sample_number,
                }

                flag_results[inpaint_methodi].append(result)

                baseline_statistics.append({
                    "baseline": str(bl),
                    "polarization": polsi,
                    "flag": flag_typesi,
                    "method": inpaint_methodi,
                    "config_number": config_idx + 1,
                    "configuration": config["label"],
                    "freq_start": config["freq_chans"].start,
                    "freq_stop": config["freq_chans"].stop,
                    "Ntime_chunks": config["Ntime_chunks"],
                    "N": sample_number,
                    "mean": mean,
                    "std": std,
                    "rmse": rmse,
                })

        # ========================================================
        # Plot each algorithm column
        # ========================================================

        for method_idx, inpaint_methodi in enumerate(inpaint_method_all):

            subplot_results = flag_results[inpaint_methodi]

            # ====================================================
            # Row 1: Histogram
            # ====================================================

            ax_hist = axes[hist_row, method_idx]

            valid_arrays = [
                result["values"]
                for result in subplot_results
                if result["values"].size > 0
            ]

            if valid_arrays:

                hist_all = np.concatenate(valid_arrays)
                hist_limit = np.quantile(np.abs(hist_all), hist_quantile)

                if not np.isfinite(hist_limit) or hist_limit <= 0:
                    hist_limit = np.max(np.abs(hist_all))

                if not np.isfinite(hist_limit) or hist_limit <= 0:
                    hist_limit = 1.0

                x_limit = hist_limit * display_fraction_x

                for result in subplot_results:

                    if result["values"].size == 0:
                        continue

                    config_idx = result["config_idx"]

                    ax_hist.hist(
                        result["values"],
                        bins=hist_bins,
                        range=(-hist_limit, hist_limit),
                        density=True,
                        histtype="step",
                        linewidth=1.8,
                        linestyle=linestyles[config_idx % len(linestyles)],
                        label=str(config_idx + 1),
                        zorder=config_idx + 1,
                    )

                ax_hist.axvline(0, color="k", linestyle="--", linewidth=0.8)
                ax_hist.set_xlim(-x_limit, x_limit)
                ax_hist.set_yscale("log")

                ax_hist.legend(
                    fontsize=6,
                    loc="upper right",
                    frameon=True,
                    handlelength=2.0,
                    labelspacing=0.2,
                    borderpad=0.3,
                    ncol=2,
                )

            else:

                ax_hist.text(
                    0.5,
                    0.5,
                    "No data",
                    ha="center",
                    va="center",
                    transform=ax_hist.transAxes,
                )

            ax_hist.tick_params(axis="both", labelsize=8)
            ax_hist.set_ylabel(f"{flag_title}\nDensity" if method_idx == 0 else "Density", fontsize=9)

            if flag_idx == 0:
                ax_hist.set_title(inpaint_methodi, fontsize=13, fontweight="bold")

            ax_hist.set_xlabel(
                r"$(|V_{\mathrm{inp}}|-|V_{\mathrm{true}}|)/|V_{\mathrm{true}}|$",
                fontsize=9,
            )

            # ====================================================
            # Row 2: RMSE versus chunks
            # Only frequency range 0–600
            # ====================================================

            ax_chunk = axes[chunk_row, method_idx]

            chunk_results = [
                result for result in subplot_results
                if result["freq_start"] == 0 and result["freq_stop"] == 600
            ]

            chunk_results = sorted(chunk_results, key=lambda result: result["Ntime_chunks"])

            chunk_x = [result["Ntime_chunks"] for result in chunk_results]
            chunk_rmse = [result["rmse"] for result in chunk_results]

            ax_chunk.plot(
                chunk_x,
                chunk_rmse,
                marker="o",
                linewidth=1.6,
                markersize=5,
            )

            ax_chunk.set_xscale("log", base=2)
            ax_chunk.set_yscale("log")

            ax_chunk.set_xticks([50, 100, 200, 400, 800])
            ax_chunk.set_xticklabels(["50", "100", "200", "400", "800"])

            ax_chunk.set_xlabel("Ntime chunks", fontsize=9)
            ax_chunk.set_ylabel("RMSE", fontsize=9)

            ax_chunk.tick_params(axis="both", labelsize=8)
            ax_chunk.grid(True, which="both", alpha=0.25)

            ax_chunk.set_title(
                "RMSE vs Chunks, Freq 0–600",
                fontsize=9,
            )

            # ====================================================
            # Row 3: RMSE versus frequency range
            # Only Ntime_chunks = 200
            # ====================================================

            ax_freq = axes[freq_row, method_idx]

            freq_configs = [
                (0, 600),
                (150, 500),
                (200, 450),
                (250, 400),
            ]

            freq_positions = np.arange(len(freq_configs))
            freq_labels = [f"{start}–{stop}" for start, stop in freq_configs]

            freq_rmse = []

            for freq_start, freq_stop in freq_configs:

                matching_results = [
                    result for result in subplot_results
                    if result["freq_start"] == freq_start
                    and result["freq_stop"] == freq_stop
                    and result["Ntime_chunks"] == 200
                ]

                if matching_results:
                    freq_rmse.append(matching_results[0]["rmse"])
                else:
                    freq_rmse.append(np.nan)

            ax_freq.plot(
                freq_positions,
                freq_rmse,
                marker="o",
                linewidth=1.6,
                markersize=5,
            )

            ax_freq.set_yscale("log")

            ax_freq.set_xticks(freq_positions)
            ax_freq.set_xticklabels(freq_labels, rotation=25, ha="right")

            ax_freq.set_xlabel("Frequency channels", fontsize=9)
            ax_freq.set_ylabel("RMSE", fontsize=9)

            ax_freq.tick_params(axis="both", labelsize=8)
            ax_freq.grid(True, which="both", alpha=0.25)

            ax_freq.set_title(
                "RMSE vs Frequency, Chunks 200",
                fontsize=9,
            )

    # ========================================================
    # Configuration legend outside the subplots
    # ========================================================

    configuration_handles = [
        Line2D(
            [0],
            [0],
            color="black",
            linewidth=1.8,
            linestyle=linestyles[i % len(linestyles)],
            label=f"{i + 1}: {config['label']}",
        )
        for i, config in enumerate(configurations)
    ]

    fig.legend(
        handles=configuration_handles,
        loc="upper center",
        bbox_to_anchor=(0.5, 0.975),
        ncol=4,
        fontsize=8,
        frameon=True,
        handlelength=2.5,
        columnspacing=1.5,
    )

    fig.suptitle(
        f"Baseline {bl}, Pol {polsi}",
        fontsize=16,
        y=0.998,
    )

    fig.subplots_adjust(
        top=0.93,
        bottom=0.03,
        left=0.08,
        right=0.98,
        hspace=0.65,
        wspace=0.30,
    )

    if save_figures:

        figure_name = (
            f"hist_rmse_grouped_bl_{bl[0]}_{bl[1]}_"
            f"pol_{polsi}.png"
        )

        fig.savefig(
            figure_dir / figure_name,
            dpi=200,
            bbox_inches="tight",
        )

    plt.show()
    plt.close(fig)

    # ========================================================
    # Statistics table
    # ========================================================

    baseline_df = pd.DataFrame(baseline_statistics)

    baseline_df["rank"] = baseline_df.groupby(
        ["flag", "method"]
    )["rmse"].rank(method="min")

    display_columns = [
        "baseline",
        "flag",
        "method",
        "config_number",
        "configuration",
        "freq_start",
        "freq_stop",
        "Ntime_chunks",
        "N",
        "mean",
        "std",
        "rmse",
        "rank",
    ]

    display(
        baseline_df[display_columns]
        .sort_values(["flag", "method", "rank"])
        .reset_index(drop=True)
    )

    # ========================================================
    # Best configuration for each flag and method
    # ========================================================

    best_configuration_df = (
        baseline_df
        .sort_values("rmse")
        .groupby(["flag", "method"], as_index=False)
        .first()
    )

    print(f"\nBest configurations for baseline {bl}:")

    display(
        best_configuration_df[
            [
                "flag",
                "method",
                "config_number",
                "configuration",
                "mean",
                "std",
                "rmse",
            ]
        ]
    )

    if save_tables:

        baseline_df.to_csv(
            table_dir / f"hist_rmse_statistics_bl_{bl[0]}_{bl[1]}_pol_{polsi}.csv",
            index=False,
        )

        best_configuration_df.to_csv(
            table_dir / f"best_rmse_configurations_bl_{bl[0]}_{bl[1]}_pol_{polsi}.csv",
            index=False,
        )

    all_statistics.append(baseline_df)


# ============================================================
# Combined table
# ============================================================

if all_statistics:

    all_statistics_df = pd.concat(all_statistics, ignore_index=True)

    all_statistics_df["overall_rank"] = all_statistics_df.groupby(
        ["baseline", "flag", "method"]
    )["rmse"].rank(method="min")

    display(
        all_statistics_df[
            [
                "baseline",
                "flag",
                "method",
                "config_number",
                "configuration",
                "freq_start",
                "freq_stop",
                "Ntime_chunks",
                "N",
                "mean",
                "std",
                "rmse",
                "overall_rank",
            ]
        ]
        .sort_values(["baseline", "flag", "method", "overall_rank"])
        .reset_index(drop=True)
    )

    if save_tables:

        all_statistics_df.to_csv(
            table_dir / f"hist_rmse_statistics_all_baselines_pol_{polsi}.csv",
            index=False,
        )
```
